```python
## Dataset: Adult (Census Income) Dataset

Dataset містить інформацію про людей з перепису населення США.
Містить як числові, так і категоріальні атрибути, наприклад:
- age
- workclass
- education
- occupation
- hours-per-week

Метою є аналіз залежностей між різними характеристиками.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"

columns = [
'age','workclass','fnlwgt','education','education-num','marital-status',
'occupation','relationship','race','sex','capital-gain','capital-loss',
'hours-per-week','native-country','income'
]

df = pd.read_csv(url, names=columns, na_values=' ?')

df.head()
```

```python
# перевірка пропущених значень
df.isnull().sum()
```

```python
# видаляю рядки з missing values
df = df.dropna()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="age", y="hours-per-week")
plt.title("Age vs Working Hours")
plt.show()
```

```python
df_sorted = df.sort_values("age")

plt.figure(figsize=(8,5))
sns.lineplot(data=df_sorted, x="age", y="hours-per-week")
plt.title("Working hours by age")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['age'], bins=5)
plt.title("Age distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="income", y="age", data=df)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="sex", data=df)
plt.show()
```
