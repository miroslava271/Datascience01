```python
### Завдання 0
Створити скрипт для завантаження VHI-даних для всіх адміністративних одиниць України (крім вибірки по всій країні, provinceID=0).  
Дані зберігати у папку `data` з додаванням дати та часу у назву файлу.  
Передбачити повторні запуски та уникати повторного завантаження.
```

```python
import os
import requests
from datetime import datetime

#папка для збереження файлів
save_folder = "data"

#список областей України
ukraine_provinces = list(range(1, 25))

#функція завантаження VHI для однієї області через requests
def download_vhi_requests(province_id, save_folder=save_folder):
    url = f"https://www.star.nesdis.noaa.gov/smcd/emb/vci/VH/get_TS_admin.php?country=UKR&provinceID={province_id}&year1=1981&year2=2024&type=Mean"
    filename = f"VHI_province{province_id}_{datetime.now().strftime('%Y%m%d_%H%M%S')}.txt"
    filepath = os.path.join(save_folder, filename)

    if not os.path.exists(filepath):
        # verify=False вимикає перевірку SSL
        r = requests.get(url, verify=False)
        with open(filepath, "wb") as f:
            f.write(r.content)
        print(f"Файл завантажено: {filename}")
    else:
        print(f"Файл для області {province_id} вже існує")
    
    return filepath

#завантажую файли для всіх областей
for province_id in ukraine_provinces:
    download_vhi_requests(province_id)
```

```python
### Завдання 1
Зчитати завантажені текстові файли у pandas DataFrame.  
Очистити дані: прибрати зайві рядки та стовпці, заповнити пропуски, додати номер області та назву.  
Підготувати DataFrame для подальшого аналізу.
```

```python
import os
import pandas as pd

data_folder = "data"

province_names = {
    1: "Вінницька", 2: "Волинська", 3: "Дніпропетровська", 4: "Донецька",
    5: "Житомирська", 6: "Закарпатська", 7: "Запорізька", 8: "Івано-Франківська",
    9: "Київська", 10: "Кіровоградська", 11: "Луганська", 12: "Львівська",
    13: "Миколаївська", 14: "Одеська", 15: "Полтавська", 16: "Рівненська",
    17: "Сумська", 18: "Тернопільська", 19: "Харківська", 20: "Херсонська",
    21: "Хмельницька", 22: "Черкаська", 23: "Чернівецька", 24: "Чернігівська"
}

files = [f for f in os.listdir(data_folder) if f.endswith(".txt")]
files.sort()

dfs = []

for file in files:
    filepath = os.path.join(data_folder, file)
    province_id = int(file.split("_")[1].replace("province", ""))
    
    #читаю файл, пропускаю перші 2 рядки
    df = pd.read_csv(filepath, sep=",", header=None, skiprows=2, engine='python')
    
    #залишаю тільки рядки з 7 колонками
    df = df[df.columns[:7]]
    
    #задаю назви колонок
    df.columns = ["year", "week", "SMN", "SMT", "VCI", "TCI", "VHI"]
    
    #приводжу числові колонки до float
    df[["SMN","SMT","VCI","TCI","VHI"]] = df[["SMN","SMT","VCI","TCI","VHI"]].apply(pd.to_numeric, errors="coerce")
    
    #додаю інформацію про область
    df["province_id"] = province_id
    df["province_name"] = province_names[province_id]
    
    #видаляю рядки без року
    df = df[pd.to_numeric(df["year"], errors="coerce").notnull()]
    
    dfs.append(df)

#об'єдную все в один DataFrame
vhi_data = pd.concat(dfs, ignore_index=True)

vhi_data.head()
```

```python
### Завдання 2
Змінити індекси областей у DataFrame `vhi_data`, щоб вони відповідали українській абетці.
```

```python
#відповідність українська абетка: index 1-24
ukr_province_order = {
    "Вінницька": 1, "Волинська": 2, "Дніпропетровська": 3, "Донецька": 4,
    "Житомирська": 5, "Закарпатська": 6, "Запорізька": 7, "Івано-Франківська": 8,
    "Київська": 9, "Кіровоградська": 10, "Луганська": 11, "Львівська": 12,
    "Миколаївська": 13, "Одеська": 14, "Полтавська": 15, "Рівненська": 16,
    "Сумська": 17, "Тернопільська": 18, "Харківська": 19, "Херсонська": 20,
    "Хмельницька": 21, "Черкаська": 22, "Чернівецька": 23, "Чернігівська": 24
}

#додаю новий стовпець з індексом за українською абеткою
vhi_data["ukr_index"] = vhi_data["province_name"].map(ukr_province_order)

vhi_data.head()
```

```python
### Завдання 3
Реалізувати функції для отримання:
1. Ряду VHI для конкретної області та року
2. Ряду VHI за діапазон років для кількох областей
3. Пошуку мінімуму, максимуму, середнього та медіани
```

```python
# 1. VHI для області за конкретний рік
def get_vhi_by_year(province_id, year):
    df = vhi_data[(vhi_data["ukr_index"] == province_id) & (vhi_data["year"] == year)]
    result = df[["week", "VHI"]]
    print(f"VHI для області {province_id} за рік {year}:")
    display(result)
    return result

# 2. VHI за діапазон років для кількох областей
def get_vhi_by_year_range(province_ids, start_year, end_year):
    df = vhi_data[(vhi_data["ukr_index"].isin(province_ids)) & 
                  (vhi_data["year"] >= start_year) & (vhi_data["year"] <= end_year)]
    result = df[["province_name", "year", "week", "VHI"]]
    print(f"VHI для областей {province_ids} з {start_year} по {end_year}:")
    display(result)
    return result

# 3. Статистика (мінімум, максимум, середнє, медіана)
def vhi_statistics(province_ids, start_year, end_year):
    df = get_vhi_by_year_range(province_ids, start_year, end_year)
    stats = df.groupby("province_name")["VHI"].agg(["min","max","mean","median"])
    print(f"Статистика VHI для областей {province_ids} з {start_year} по {end_year}:")
    display(stats)
    return stats

#перетворюю колонку year у int
vhi_data["year"] = pd.to_numeric(vhi_data["year"], errors="coerce")

#приклади виклику функцій

# 1. VHI для області 10 (Кіровоградська) за 1985 рік
get_vhi_by_year(10, 1985)

# 2. VHI для областей 10, 12, 15 за 1985-1986 роки
get_vhi_by_year_range([10, 12, 15], 1985, 1986)

# 3. Статистика VHI для областей 10, 12, 15 за 1985-1986 роки
vhi_statistics([10, 12, 15], 1985, 1986)
```
